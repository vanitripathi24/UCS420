```python
##q2 b part
import numpy as np
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values, counts = np.unique(x, return_counts=True)
max_count = counts.max()
most_frequent_values = values[counts == max_count]

print("original array : ", x)
print("most frequent values : ", most_frequent_values)
print("Frequency:", max_count)

for value in most_frequent_values:
    indices = np.where(x == value)[0]
    print("Indices:", indices)
```

```python
##q5
import numpy as np
ucs420_Vani = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_Vani)

print("\nMean:", np.mean(ucs420_Vani))
print("Median:", np.median(ucs420_Vani))
print("Standard Deviation:", np.std(ucs420_Vani))
print("Variance:", np.var(ucs420_Vani))
print("Sum:", np.sum(ucs420_Vani))
print("Product:", np.prod(ucs420_Vani))
print("Minimum:", np.min(ucs420_Vani))
print("Maximum:", np.max(ucs420_Vani))

reshape_ucs420_Vani = ucs420_Vani.reshape(4, 3)
print("\nReshaped Array:")
print(reshape_ucs420_Vani)

resized_ucs420_Vani = np.resize(ucs420_Vani, (2, 3))

print("\nResized Array:")
print(resized_ucs420_Vani)
```
