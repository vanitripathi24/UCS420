```python
##q1
import numpy as np
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print("original temp : ", temperature)

corrected_temp = temperature + 2
print("corrected temp : ", corrected_temp)

fahrenheit = (9 / 5) * corrected_temp + 32
print("\nTemperature in Fahrenheit:", fahrenheit)

reading_greater_32 = corrected_temp[corrected_temp > 32]
print("greater than 32 : " ,reading_greater_32)

count = len(reading_greater_32)
print("count : ", count)
```

```python
##q2

import numpy as np

steps = np.array([ [5000, 6200, 7100], [8000, 7500, 9000], [4500, 5100, 4800], [9000, 8500, 9500]])
print("total steps : ",np.sum(steps) )
print("mean steps : ", np.mean(steps))
print("median steps : ", np.median(steps))
print("standard deviation : ", np.std(steps))
print("max steps : ", np.max(steps))
print("min steps : ", np.min(steps))
print("Total steps for each day:", np.sum(steps, axis=0))
print("Total steps for each user:", np.sum(steps, axis=1))

max_position = np.argmax(steps)
print("max position : ", max_position)

min_position = np.argmin(steps)
print("min position : ", min_position)
```
