```python
import random
roll_number = int(input("Enter your roll number: "))
random.seed(roll_number)
numbers = [random.randint(100, 900) for _ in range(100)]
print("List of 100 random numbers:")
print(numbers)
odd_numbers = [num for num in numbers if num % 2 != 0]
print("\nOdd numbers:")
print(odd_numbers)
print("Number of odd numbers:", len(odd_numbers))
even_numbers = [num for num in numbers if num % 2 == 0]

print("\nEven numbers:")
print(even_numbers)
print("Number of even numbers:", len(even_numbers))
def is_prime(num):
    if num < 2:
        return False

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False

    return True
prime_numbers = [num for num in numbers if is_prime(num)]

print("\nPrime numbers:")
print(prime_numbers)
print("Number of prime numbers:", len(prime_numbers))

most_frequent = max(set(numbers), key=numbers.count)
frequency = numbers.count(most_frequent)

print("\nMost frequently occurring number:", most_frequent)
print("Number of times it occurs:", frequency)
```

```python
roll_no = input("Enter your roll number: ")
digits = [int(x) for x in roll_no[:8]]

A = {x * 7 for x in digits}
B = {x * 9 for x in digits}
print("set A : ", A)
print("set B : ", B)

print(A.union(B))
print(A.intersection(B))

print(A.difference(B))
print(B.difference(A))

print(A.symmetric_difference(B))
print(B.symmetric_difference(A))

print(A.issubset(B))
print(B.issuperset(A))

X = int(input("Enter a value to remove from A: "))
A.discard(X)
print(A)
```
