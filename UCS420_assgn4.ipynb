```python
import pandas as pd

# Sample data
data = {
    "question": [
        "how to reset password",
        "what is the annual fee",
        "how can i pay the fee",
        "what are your working hours"
    ],
    "answer": [
        "Go to Settings > Reset Password.",
        "The annual fee is Rs 500.",
        "You can pay via UPI, card, or net banking.",
        "We are open 9 AM to 5 PM."
    ],
    "keywords": [
        "password reset login",
        "fee cost price charge",
        "pay payment upi fee",
        "hours timing open time"
    ]
}

df = pd.DataFrame(data)


def score_query(query):
    query_words = set(query.lower().split())
    results = []

    for _, row in df.iterrows():

        text = row["question"] + " " + row["keywords"]
        words = set(text.lower().split())

        matched_words = query_words.intersection(words)

        if len(matched_words) > 0:
            confidence = len(matched_words) / len(query_words)

            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "confidence": confidence
            })

    # Rank by confidence
    results.sort(key=lambda x: x["confidence"], reverse=True)

    return pd.DataFrame(results)


query = input("Enter your query: ")

result = score_query(query)

print("\nMatching entries:")
print(result)
```

```python
def score_query(query):
    query_words = set(query.lower().split())
    results = []

    for _, row in df.iterrows():

        text = row["question"].lower() + " " + row["keywords"].lower()
        entry_words = set(text.split())

        matched_words = query_words.intersection(entry_words)

        if len(matched_words) > 0:
            confidence = len(matched_words) / len(query_words)

            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": confidence
            })

    # Convert to DataFrame
    results_df = pd.DataFrame(results)

    if results_df.empty:
        print("No matching entries found.")
        return results_df

    # Find highest score
    highest_score = results_df["confidence"].max()

    # Return ALL entries having the highest score
    best_matches = results_df[
        results_df["confidence"] == highest_score
    ]

    return best_matches
```
